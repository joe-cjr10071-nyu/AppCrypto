# Project 2 — One-Time Pad: Bob's Decryption

This notebook implements Bob's portion of the one-time-pad exercise.

Bob:
1. Reads the ciphertext from a hexadecimal text file.
2. Reads the key from a hexadecimal text file.
3. Converts both values to bytes.
4. XORs the ciphertext with the key.
5. Decodes and displays the recovered plaintext.

```python
from pathlib import Path
```

## Functions: Get ciphertext & key

The following two code blocks are the XOR function `xor_bytes()` which contains the exception handler which evaluates if the ciphertex and key are non-matching in length, and the `read_hex_file()` to read in the hexadecimal content and return the bytecode.

```python
def xor_bytes(data: bytes, key: bytes) -> bytes:
    if len(data) != len(key):
        raise ValueError("Ciphertext and key must have the same length.")

    return bytes(d ^ k for d, k in zip(data, key))
```

```python
def read_hex_file(filename: str) -> bytes:
    hex_data = Path(filename).read_text(
        encoding="ASCII"
    ).strip()

    return bytes.fromhex(hex_data)
```

## Bob's Decrypt & Read

This section will read in the ciphertext via and the key from hexadecimal encoding and XOR the stream to decrypt and display Alice's message. `read_hex_file` and decodes the XOR'ed hexadecimal to ASCII; it will then output the following proofs:

* ciphertext in hexadecimal
* key in hexadecimal
* Alice's plaintext 

```python
ciphertext = read_hex_file("ciphertext.txt")
key = read_hex_file("key.txt")

plaintext_bytes = xor_bytes(ciphertext, key)
plaintext = plaintext_bytes.decode("ascii")

print("Ciphertext hex:", ciphertext.hex())
print("Key hex:       ", key.hex())
print("Plaintext:     ", plaintext)
```
