# Project 2 — One-Time Pad: Bob Decryption

This notebook implements Bob's portion of the one-time-pad exercise.

Bob:
1. Reads the ciphertext from a hexadecimal text file.
2. Reads the key from a hexadecimal text file.
3. Converts both values to bytes.
4. XORs the ciphertext with the key.
5. Decodes and displays the recovered plaintext.

```python
from pathlib import Path


def xor_bytes(message: bytes, key: bytes) -> bytes:
    if len(message) != len(key):
        raise ValueError("Ciphertext and key must be the same length.")

    return bytes(m ^ k for m, k in zip(message, key))


def read_hex_file(filename: str) -> bytes:
    hex_text = Path(filename).read_text(encoding="utf-8").strip()
    return bytes.fromhex(hex_text)


def decrypt(ciphertext: bytes, key: bytes) -> str:
    plaintext_bytes = xor_bytes(ciphertext, key)
    return plaintext_bytes.decode("utf-8")
```

```python
ciphertext = read_hex_file("ciphertext.txt")
key = read_hex_file("key.txt")

plaintext = decrypt(ciphertext, key)

print("Recovered plaintext:")
print(plaintext)
```
